```python
print("Arun Kumar N S")
print("1WN24CS050")
print()
SUBST = {}

def is_variable(x):
    return isinstance(x, str) and x[0].islower()


def apply_substitution(x):
    if isinstance(x, str):
        if is_variable(x) and x in SUBST:
            return apply_substitution(SUBST[x])
        return x

    return (x[0], tuple(apply_substitution(a) for a in x[1]))


def occurs(var, term):
    term = apply_substitution(term)

    if isinstance(term, str):
        return var == term

    return any(occurs(var, arg) for arg in term[1])


def unify(x, y):
    x = apply_substitution(x)
    y = apply_substitution(y)

    # Step 1: Check variables and constants
    if isinstance(x, str) and isinstance(y, str):
        if x == y:
            return True

        if is_variable(x):
            if occurs(x, y):
                return False
            SUBST[x] = y
            return True

        if is_variable(y):
            if occurs(y, x):
                return False
            SUBST[y] = x
            return True

        return False

    if isinstance(x, str):
        if is_variable(x):
            if occurs(x, y):
                return False
            SUBST[x] = y
            return True
        return False

    if isinstance(y, str):
        if is_variable(y):
            if occurs(y, x):
                return False
            SUBST[y] = x
            return True
        return False

    # Step 2: Check predicate symbols
    if x[0] != y[0]:
        return False

    # Step 3: Check number of arguments
    if len(x[1]) != len(y[1]):
        return False

    # Step 4: Initialize substitution set
    # Step 5: Unify corresponding arguments
    for a, b in zip(x[1], y[1]):
        if not unify(a, b):
            return False

    return True


def parse(expression):
    expression = expression.strip()
    pos = expression.find('(')

    if pos == -1:
        return expression

    name = expression[:pos].strip()
    inside = expression[pos + 1:-1]

    args = []
    start = 0
    depth = 0

    for i, ch in enumerate(inside):
        if ch == '(':
            depth += 1
        elif ch == ')':
            depth -= 1
        elif ch == ',' and depth == 0:
            args.append(parse(inside[start:i]))
            start = i + 1

    args.append(parse(inside[start:]))

    return (name, tuple(args))


def display(term):
    if isinstance(term, str):
        return term

    return term[0] + "(" + ", ".join(
        display(arg) for arg in term[1]
    ) + ")"


# Main Program
e1 = input("Enter first expression: ")
e2 = input("Enter second expression: ")

A = parse(e1)
B = parse(e2)

SUBST.clear()

if unify(A, B):
    print("\nUnification Successful")
    print("Substitution Set:")

    if SUBST:
        for var, value in SUBST.items():
            print(var, "=", display(apply_substitution(value)))
    else:
        print("NIL")

    print("Unified Expression:", display(apply_substitution(A)))

else:
    print("\nUnification Failed")
```
