8 Puzzle DFS

```python
print("Arun Kumar N S")
print("1WN24CS050")
def dfs(state, goal, visited, path, cost):

    if state == goal:
        return path, cost

    visited.add(tuple(state))

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = state.copy()

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            if tuple(new_state) not in visited:

                result = dfs(
                    new_state,
                    goal,
                    visited,
                    path + [new_state],
                    cost + 1
                )

                if result:
                    return result

    return None


initial = [1, 2, 3,
           0, 4, 6,
           7, 5, 8]

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

visited = set()

result = dfs(initial, goal, visited, [initial], 0)

if result:

    path, cost = result

    print("Solution Path:")

    for state in path:
        print(state[0:3])
        print(state[3:6])
        print(state[6:9])
        print()

    print("Path Cost:", cost)

else:
    print("No solution")
```

IDS

```python
print("Arun Kumar N S")
print("1WN24CS050")
def depth_limited_dfs(state, goal, depth, path, cost):

    if state == goal:
        return path, cost

    if depth == 0:
        return None

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = state.copy()

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            if new_state not in path:

                result = depth_limited_dfs(
                    new_state,
                    goal,
                    depth - 1,
                    path + [new_state],
                    cost + 1
                )

                if result:
                    return result

    return None


def ids(initial, goal):

    depth = 0

    while True:

        result = depth_limited_dfs(
            initial,
            goal,
            depth,
            [initial],
            0
        )

        if result:
            return result

        depth += 1


initial = [1, 2, 3,
           0, 4, 6,
           7, 5, 8]

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

result = ids(initial, goal)

if result:

    path, cost = result

    print("Solution Path:")

    for state in path:
        print(state[0:3])
        print(state[3:6])
        print(state[6:9])
        print()

    print("Path Cost:", cost)

else:
    print("No solution")
```
